```python
import pandas as pd
import altair as alt

# Load the Brent Oil dataset
data = pd.read_csv(r'C:\Users\Disha\IISER - B Summer Internship\Brent Oil Futures Historical Data.csv')

# Convert the 'Date' column to datetime format
data['Date'] = pd.to_datetime(data['Date'])

# Create a line chart for 'Price' over time
chart = alt.Chart(data).mark_line().encode(
    x='Date:T',  # 'T' indicates that it's a time axis
    y='Price:Q',  # 'Q' indicates that it's a quantitative axis (numeric values)
    color=alt.value('steelblue')  # Line color
).properties(
    title='Brent Oil Futures Price over Time',
    width=800,
    height=400
)

# Show the chart in the notebook
chart.show()
```
