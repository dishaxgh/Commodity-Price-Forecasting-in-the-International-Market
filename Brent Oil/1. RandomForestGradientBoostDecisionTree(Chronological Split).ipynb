```python
# extract features and target
X = data[['Open', 'High', 'Low','Vol.']]
y = data['Price']
```

```python
np.random.seed(42)
```

```python
"""
from sklearn.model_selection import train_test_split

# Split the data into a 80/20 train/test split
train_X, test_X, train_y, test_y = train_test_split(data, y, test_size=0.2, random_state=42)

"""
```

```python
# Addition
data['Date'] = pd.to_datetime(data['Date'], format='%Y-%m-%d')
```

```python
# Addition
# Select the rows for the training set (2000-2019)
train_X = X[data['Date'].dt.year.between(2000, 2019)]
train_y = y[data['Date'].dt.year.between(2000, 2019)]

# Select the rows for the test set (2020-2022)
test_X = X[data['Date'].dt.year.between(2020, 2022)]
test_y = y[data['Date'].dt.year.between(2020, 2022)]
```

```python
# Fitting Random Forest Regressor to the training set  
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

# Create a Random Forest Regressor
regressor = RandomForestRegressor(n_estimators=200, max_leaf_nodes=10, max_depth=10, random_state=42)

# Train the model using the training data
regressor.fit(train_X, train_y)

# Use the fitted model to make predictions on the test data
pred = regressor.predict(test_X)

# Calculate R-squared
r2 = r2_score(test_y,pred)
print(r2)

# Evaluate the model's performance
from sklearn.metrics import mean_squared_error
rmse = np.sqrt(mean_squared_error(test_y, pred))
print("Root Mean Squared Error: ", rmse)
```

```python
# Plot actual vs predicted values
plt.scatter(test_y, pred)
plt.xlabel("Actual values")
plt.ylabel("Predicted values")
plt.title(f"Random Forest Regressor (R-squared={r2:.2f})")
plt.show()
```

```python
# Gradient Boost
from sklearn.ensemble import GradientBoostingRegressor

# Initialize the gradient boost regressor
gb = GradientBoostingRegressor(n_estimators=300, max_leaf_nodes=10, max_depth=10, random_state=42, max_features=3)

# Fit the regressor to the training data
gb.fit(train_X, train_y)

# Make predictions on the test data
pred = gb.predict(test_X)

# Calculate R-squared
r2 = r2_score(test_y,pred)
print(r2)

# Evaluate the model's performance
from sklearn.metrics import mean_squared_error
rmse = np.sqrt(mean_squared_error(test_y, pred))
print("Root Mean Squared Error: ", rmse)
```

```python
# Plot actual vs predicted values
plt.scatter(test_y, pred)
plt.xlabel("Actual values")
plt.ylabel("Predicted values")
plt.title(f"Gradient Boost (R-squared={r2:.2f})")
plt.show()
```

```python
# Decision Tree Regressor
from sklearn.tree import DecisionTreeRegressor

# Create an instance of DecisionTreeRegressor with max depth of 5
dt_regressor = DecisionTreeRegressor(max_depth=3, random_state=42, max_features=3)

# Fit the model on training data
dt_regressor.fit(train_X, train_y)

# Make predictions on test data
pred = dt_regressor.predict(test_X)

# Calculate R-squared
r2 = r2_score(test_y,pred)
print(r2)

# Evaluate the model's performance
from sklearn.metrics import mean_squared_error
rmse = np.sqrt(mean_squared_error(test_y, pred))
print("Root Mean Squared Error: ", rmse)
```

```python
# Plot actual vs predicted values
plt.scatter(test_y, pred)
plt.xlabel("Actual values")
plt.ylabel("Predicted values")
plt.title(f"Decision Tree (R-squared={r2:.2f})")
plt.show()
```

```python
# Stacking Ensemble model
from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# Define the base models
gradient_boost_model = VotingRegressor([('gradient_boost_1', gradient_boost_1),
                                        ('gradient_boost_2', gradient_boost_2),
                                        ('gradient_boost_3', gradient_boost_3)])

bagging_model = VotingRegressor([('random_forest', regressor),
                                 ('decision_tree', dt_regressor)])
```

```python
# Define the meta-model
meta_model = LinearRegression()
```

```python
# Create the stacking ensemble
ensemble = StackingRegressor(estimators=[('gradient_boost', gradient_boost_model),
                                         ('bagging', bagging_model)],
                                          final_estimator=meta_model)
```

```python
# Train the stacking ensemble on the training data
ensemble.fit(train_X, train_y)
```

```python
# Make predictions using the stacking ensemble
ensemble_preds = ensemble.predict(test_X)
```

```python
# Evaluate the stacking ensemble
# Calculate R-squared
r2 = r2_score(test_y,ensemble_preds)
print(r2)

ensemble_rmse = mean_squared_error(test_y, ensemble_preds, squared=False) 
print("Ensemble RMSE:", ensemble_rmse)
```

```python
# Plot actual vs predicted values
plt.scatter(test_y, ensemble_preds)
plt.xlabel("Actual values")
plt.ylabel("Predicted values")
plt.title(f"Ensemble Model (R-squared={r2:.2f})")
plt.show()
```
