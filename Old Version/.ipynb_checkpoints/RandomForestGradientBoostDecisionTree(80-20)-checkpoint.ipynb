```python
# Importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Reading the file
import pandas as pd
data = pd.read_csv(r"C:\Users\Disha\IISER - B Summer Internship\Brent Oil Futures Historical Data.csv")
```

```python
# Print the first 5 rows of the dataset
print(data.head())

# Get a summary of the dataset
print(data.dtypes)

# Get summary statistics for numerical columns
print(data.describe())
```

```python
# Converting "Vol." column to float datatype
data['Vol.'] = data['Vol.'].str.replace('K', '').astype(float) * 1000
print(data.dtypes)
```

```python
# Check for null values in each column
null_counts = data.isnull().sum()

# Print the number of null values in each column
print(null_counts)
```

```python
# fill missing values with mean of the column
data = data.fillna(data.mean())

# Check for null values in each column again
null_counts = data.isnull().sum()

# Print the number of null values in each column
print(null_counts)
```

```python
# Create a boxplot of a numerical column
sns.boxplot(data['Price'])
plt.show()
```

```python
# Drop a column from the dataset
data = data.drop('Change %', axis=1)

# Show the updated dataset
print(data.head())
```

```python
# Convert the "date" column to a datetime object
data['Date'] = pd.to_datetime(data['Date'])

# Create a line graph of date and price
plt.plot(data['Date'], data['Price'])

# Add x-label, y-label, and title to the plot
plt.xlabel('Date')
plt.ylabel('Price')
plt.title('Price over Time')

# Show the plot
plt.show()
```

```python
# Convert the "date" column to a datetime object
data['Date'] = pd.to_datetime(data['Date'])

# Create a line graph of date and Open price
plt.plot(data['Date'], data['Open'])

# Add x-label, y-label, and title to the plot
plt.xlabel('Date')
plt.ylabel('Opening Price')
plt.title('Opening Price over Time')

# Show the plot
plt.show()
```

```python
# Convert the "date" column to a datetime object
data['Date'] = pd.to_datetime(data['Date'])

# Create a line graph of date and high price
plt.plot(data['Date'], data['High'])

# Add x-label, y-label, and title to the plot
plt.xlabel('Date')
plt.ylabel('Highest Price')
plt.title('Highest Price over Time')

# Show the plot
plt.show()
```

```python
# Convert the "date" column to a datetime object
data['Date'] = pd.to_datetime(data['Date'])

# Create a line graph of date and low price
plt.plot(data['Date'], data['Low'])

# Add x-label, y-label, and title to the plot
plt.xlabel('Date')
plt.ylabel('Lowest Price')
plt.title('Lowest Price over Time')

# Show the plot
plt.show()
```

```python
# Create a scatter plot of two numerical columns
plt.scatter(data['Date'], data['Price'])

# Add labels and title to the plot
plt.xlabel('Date')
plt.ylabel('Price')
plt.title('Scatter Plot of Date vs Price')

# Show the plot
plt.show()
```

```python
# Drop a column from the dataset
data = data.drop('Date', axis=1)

# Show the updated dataset
print(data.head())
```

```python
# extract features and target
X = data[['Open', 'High', 'Low','Vol.']]
y = data['Price']
```

```python
from sklearn.model_selection import train_test_split

# Split the data into a 80/20 train/test split
train_X, test_X, train_y, test_y = train_test_split(data, y, test_size=0.2, random_state=42)
```

```python
# Fitting Random Forest Regressor to the training set  
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

# Create a Random Forest Regressor
regressor = RandomForestRegressor(n_estimators=100, random_state=42)

# Train the model using the training data
regressor.fit(train_X, train_y)

# Use the fitted model to make predictions on the test data
pred = regressor.predict(test_X)

# Calculate R-squared
r2 = r2_score(test_y,pred)
print(r2)

# Evaluate the model's performance
from sklearn.metrics import mean_squared_error
rmse = np.sqrt(mean_squared_error(test_y, pred))
print("Root Mean Squared Error: ", rmse)
```

```python
# Plot actual vs predicted values
plt.scatter(test_y, pred)
plt.xlabel("Actual values")
plt.ylabel("Predicted values")
plt.title(f"Random Forest Regressor (R-squared={r2:.2f})")
plt.show()
```

```python
# Gradient Boost
from sklearn.ensemble import GradientBoostingRegressor

# Initialize the gradient boost regressor
gb = GradientBoostingRegressor()

# Fit the regressor to the training data
gb.fit(train_X, train_y)

# Make predictions on the test data
pred = gb.predict(test_X)

# Calculate R-squared
r2 = r2_score(test_y,pred)
print(r2)

# Evaluate the model's performance
from sklearn.metrics import mean_squared_error
rmse = np.sqrt(mean_squared_error(test_y, pred))
print("Root Mean Squared Error: ", rmse)
```

```python
# Plot actual vs predicted values
plt.scatter(test_y, pred)
plt.xlabel("Actual values")
plt.ylabel("Predicted values")
plt.title(f"Gradient Boost (R-squared={r2:.2f})")
plt.show()
```

```python
# Decision Tree Regressor
from sklearn.tree import DecisionTreeRegressor

# Create an instance of DecisionTreeRegressor with max depth of 5
dt_regressor = DecisionTreeRegressor(max_depth=5)

# Fit the model on training data
dt_regressor.fit(train_X, train_y)

# Make predictions on test data
pred = dt_regressor.predict(test_X)

# Calculate R-squared
r2 = r2_score(test_y,pred)
print(r2)

# Evaluate the model's performance
from sklearn.metrics import mean_squared_error
rmse = np.sqrt(mean_squared_error(test_y, pred))
print("Root Mean Squared Error: ", rmse)
```

```python
# Plot actual vs predicted values
plt.scatter(test_y, pred)
plt.xlabel("Actual values")
plt.ylabel("Predicted values")
plt.title(f"Decision Tree (R-squared={r2:.2f})")
plt.show()
```
